task 1 -> create a dictionary with named employees with 2 employees (emp_1,emp_2).
each employee should have: 
-name (string)
-age(int)
_department (str)

**Nested structure inside each employee :**
1 salary_details (dictionary)
base_salary
bonus
projects (dictionary)
project_A -status (completed/In progress)
project_B -status 

```python
# Employee Folder with two employees
employee={
  "emp_1" : {
  "name":"avic",
  "age": 25,
  "department": "IT",
  "salary_details":{
    "base_salary": 100000,
    "bonus": 5000
  },
  "project_A": "Completed",
  "project_B": "pending",
},
 "emp_2": {
  "name":"David",
  "age": 27,
  "department": "finance",
  "salary_details":{
    "base_salary": 10000,
    "bonus": 500
  },
  "project_A": "Completed",
  "project_B": "N/A",
}        }
```

```python
#task 1 Access values print :
print(employee["emp_1"]["name"],employee["emp_2"]["department"])
```

```python
#task.2 Access nested data print : 
print(employee["emp_1"]["salary_details"]["bonus"],employee["emp_2"]["project_B"])

#update values
employee["emp_1"]["salary_details"]["base_salary"]=55000
employee["emp_2"]["project_B"]="Completed"

#add new data 
employee["emp_1"]["project_C"]="Completed"
employee["emp_2"]["experience"]=3
print(employee)
```
