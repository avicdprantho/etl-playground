## Step 6: Add or Modify Columns

```python
from pyspark.sql.functions import col
df=df.withColumn("profit", col("amount")- col("price"))
```

## Step 7: Rename Columns

```python
df=df.withColumnRenamed("profit", "delete")
df.display()
```
